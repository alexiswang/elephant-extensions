```python
from pathlib import Path
import pandas as pd

from analyze_experiment_results import (
    calculate_sycophancy,
    compare_conditions,
    pair_level_table,
    format_metric_table
)
```

```python
r1_baseline_path = Path("results/model_evaluations/deepseek_deepseek-r1-0528_baseline")
r1_no_nudge_path = Path("results/model_evaluations/deepseek_deepseek-r1-0528_baseline")
r1_nta_nudge_path = Path("results/model_evaluations/deepseek_deepseek-r1-0528_baseline")
r1_yta_nudge_path = Path("results/model_evaluations/deepseek_deepseek-r1-0528_baseline")
v3_baseline_path = Path("results/model_evaluations/deepseek_deepseek-chat_baseline")
```

# Compare V3 and R1

```python
replication_table = format_metric_table(
    {
        "V3 (n=400)": calculate_sycophancy(
            pair_level_table(v3_baseline_path)
        ),
        "R1 (n=400)": calculate_sycophancy(
            pair_level_table(r1_baseline_path)
        ),
    }
)

print(replication_table.to_markdown(index=False))
```

# Compare sycophancy rate of R1 across data fidelity groups

```python
# Filter all v3 baseline pairs into high and low fidelity subsets
fidelity_results = pd.read_csv("results/fidelity_audit/analysis/pivoted_verdicts.csv")

high_fidelity_ids = fidelity_results.loc[
    fidelity_results["fidelity_verdict"] == "No",
    "dataset_index",
]

v3_baseline_pairs = pair_level_table(r1_baseline_path)
v3_high_fidelity_pairs = v3_baseline_pairs[
    v3_baseline_pairs["dataset_index"].isin(high_fidelity_ids)
].copy()

v3_low_fidelity_pairs = v3_baseline_pairs[
    ~v3_baseline_pairs["dataset_index"].isin(high_fidelity_ids)
].copy()

v3_high_fidelity_stats = calculate_sycophancy(v3_high_fidelity_pairs)
v3_low_fidelity_stats = calculate_sycophancy(v3_low_fidelity_pairs)
```

```python
fidelity_metrics = format_metric_table(
    {
        "High Fidelity": v3_high_fidelity_stats,
        "Low Fidelity": v3_low_fidelity_stats,
    }
)

print(fidelity_metrics.to_markdown(index=False))
```

# Compare sycophancy across different nudge settings

```python
nudge_metrics = format_metric_table(
    {
        "Baseline": calculate_sycophancy(
            pair_level_table(r1_baseline_path)
        ),
        "No nudge": calculate_sycophancy(
            pair_level_table(r1_no_nudge_path)
        ),
        "NTA nudge": calculate_sycophancy(
            pair_level_table(r1_nta_nudge_path)
        ),
        "YTA nudge": calculate_sycophancy(
            pair_level_table(r1_yta_nudge_path)
        ),
    }
)

print(nudge_metrics.to_markdown(index=False))
```
